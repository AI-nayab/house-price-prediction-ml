```python
df.drop(columns=['society'],inplace=True)
```

```python
df['location']=df['location'].apply(lambda x:x.strip())
```

```python
df.location.value_counts()
```

```python
location_stats=df.groupby('location')['location'].agg('count').sort_values(ascending=False)
```

```python
df.location.value_counts()
```

```python
location_stats
```

```python
location_lessthan_10_entries=location_stats[location_stats<=10]
```

```python
location_lessthan_10_entries
```

```python
df['location']=df['location'].apply(lambda x:'other' if x in location_lessthan_10_entries else x)
```

```python
df['location'].value_counts()
```

```python
df
```

```python
df['size'].unique()
```

```python
df['bedrooms']=df['size'].apply(lambda x:int(x.split(' ')[0]))
```

```python
df
```

```python
df.	total_sqft.unique()
```

```python
def clean(sqft):
    tokens = sqft.split('-')
    if len(tokens) == 2:
        return (float(tokens[0]) + float(tokens[1]))/2
    else:
        try:
            return float(sqft)
        except:
            return None
```

```python
df['total_sqft']=df['total_sqft'].apply(clean)
```

```python
df['total_sqft'].unique()
```

```python
df.dropna(inplace=True)
```

```python
df.describe()
```

```python
df['sqft_per_bedroom']=df['total_sqft']/df['bedrooms']
```

```python
df.sqft_per_bedroom.describe()
```

```python
df2=df[df['sqft_per_bedroom'] >=300]
```

```python
df2
```

```python
df2['price_per_sqft'] = round(df2['price']*100000/df2['total_sqft'],2)
```

```python
df2.price_per_sqft.describe()
```

```python
df3 = df2 [df2['price_per_sqft'] >= 2000]
```

```python
df3
```

```python
df3.drop(columns = ['size', 'sqft_per_bedroom', 'price_per_sqft'])
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.pipeline import make_pipeline
```

```python
from sklearn.compose import make_column_transformer

```

```python
col_trans = make_column_transformer(
    (
        OneHotEncoder(sparse_output=False, handle_unknown='ignore'),
        ['area_type', 'availability', 'location', 'size']
    ),
    remainder='passthrough'
)
```

```python
lr = LinearRegression()
```

```python
scaler = StandardScaler()
```

```python
pipe = make_pipeline(
    col_trans,
    StandardScaler(with_mean=False),
    LinearRegression()
)
```

```python
data_input = df3.drop(columns=['price'])
data_output = df3['price']
```

```python
x_train, x_test, y_train, y_test = train_test_split(data_input, data_output, test_size=0.2)
```

```python
pipe.fit(x_train, y_train)
```

```python
pipe.score(x_test, y_test)
```
